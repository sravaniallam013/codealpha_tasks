## Task 1: Web Scraping using BeautifulSoup

### Objective
Extract Quotes and Author names from https://quotes.toscrape.com and save them into a CSV file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Step 1: Fetch the Website

```python
url = "https://quotes.toscrape.com"

response = requests.get(url)

print(response.status_code)
```

## Step 2: Parse the HTML Content

```python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)
```

## Step 3: Extract Quotes and Authors

```python
quotes = soup.find_all("div", class_="quote")

quote_list = []
author_list = []

for q in quotes:
    text = q.find("span", class_="text").text
    author = q.find("small", class_="author").text

    quote_list.append(text)
    author_list.append(author)

print("Total Quotes:", len(quote_list))
```

## Step 4: Create a DataFrame

```python
df = pd.DataFrame({
    "Quote": quote_list,
    "Author": author_list
})

df.head()
```

## Step 5: Save Data to CSV File

```python
df.to_csv("quotes.csv", index=False)

print("✅ quotes.csv file created successfully!")
```
